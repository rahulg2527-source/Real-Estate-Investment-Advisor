```python
price_check = pd.DataFrame()

price_check["Actual_Price_per_SqFt"] = (
    df["Price_per_SqFt"]
)

price_check["Calculated_Price_per_SqFt"] = (
    df["Price_in_Lakhs"] * 100000
    / df["Size_in_SqFt"]
)

price_check["Difference"] = (
    price_check["Actual_Price_per_SqFt"]
    - price_check["Calculated_Price_per_SqFt"]
)

price_check["Percentage_Error"] = (
    price_check["Difference"].abs()
    / price_check["Calculated_Price_per_SqFt"].abs()
    * 100
)

display(price_check.describe().T)
```

```python
# Compare actual price per sqft with calculated price

price_check = pd.DataFrame({
    "Actual": df["Price_per_SqFt"],
    "Calculated_Rupees": (
        df["Price_in_Lakhs"] * 100000
        / df["Size_in_SqFt"]
    )
})

price_check["Converted_Actual"] = (
    price_check["Actual"] * 100000
)

price_check["Difference"] = (
    price_check["Converted_Actual"]
    - price_check["Calculated_Rupees"]
)

price_check["Relative_Error_Percent"] = (
    price_check["Difference"].abs()
    / price_check["Calculated_Rupees"].abs()
) * 100

display(price_check.head(15))
display(price_check["Relative_Error_Percent"].describe())
```

```python
# ============================================
# REAL ESTATE INVESTMENT ADVISOR
# DAY 2 - DATA CLEANING
# ============================================

import os
import pandas as pd
import numpy as np

# Always restart from original data
df_clean = df_raw.copy(deep=True)

# 1. Remove exact duplicate rows (if any)
df_clean = df_clean.drop_duplicates().copy()

# 2. Preserve original price-per-sqft column
df_clean.rename(
    columns={
        "Price_per_SqFt": "Price_per_SqFt_Original_Lakhs"
    },
    inplace=True
)

# 3. Calculate accurate price per sqft in INR
df_clean["Price_per_SqFt"] = (
    df_clean["Price_in_Lakhs"] * 100000
    / df_clean["Size_in_SqFt"]
)

# 4. Convert Yes/No columns to binary
binary_columns = [
    "Parking_Space",
    "Security"
]

for col in binary_columns:
    df_clean[col] = (
        df_clean[col]
        .astype(str)
        .str.strip()
        .str.lower()
        .map({"yes": 1, "no": 0})
        .astype("int8")
    )

# 5. Extract individual amenities
amenities = [
    "Pool",
    "Clubhouse",
    "Garden",
    "Gym",
    "Playground"
]

for amenity in amenities:
    column_name = f"Has_{amenity}"

    df_clean[column_name] = (
        df_clean["Amenities"]
        .fillna("")
        .apply(
            lambda x: int(
                amenity in [
                    item.strip()
                    for item in x.split(",")
                ]
            )
        )
    )

# 6. Preserve floor information but don't use
# these columns in initial ML models.
# Their reliability is questionable.

# 7. Save cleaned dataset
OUTPUT_PATH = (
    "../data/processed/cleaned_housing_prices.csv"
)

os.makedirs(
    "../data/processed",
    exist_ok=True
)

df_clean.to_csv(
    OUTPUT_PATH,
    index=False
)

print("Cleaning completed successfully!")
print("Final shape:", df_clean.shape)
print("Saved to:", OUTPUT_PATH)
```

```python
# Reload saved processed dataset
df_verified = pd.read_csv(OUTPUT_PATH)

print("Original records:", len(df_raw))
print("Cleaned records:", len(df_verified))

print("Missing values:", df_verified.isnull().sum().sum())
print("Duplicate rows:", df_verified.duplicated().sum())

print("\nNew columns:")
print(df_verified.columns.tolist())

print("\nDataset saved and verified!")
```

```python
# Flag zero values in the original reported column

df_clean["Original_Price_per_SqFt_Was_Zero"] = (
    df_clean["Price_per_SqFt_Original_Lakhs"] == 0
).astype("int8")

# Save updated processed dataset

df_clean.to_csv(
    OUTPUT_PATH,
    index=False
)

# Verify
print(
    "Zero-value records flagged:",
    df_clean["Original_Price_per_SqFt_Was_Zero"].sum()
)

print("Final dataset shape:", df_clean.shape)
print("Dataset saved successfully!")
```
